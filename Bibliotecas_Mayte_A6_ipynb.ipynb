```python
# 19

import numpy as np

# Gera dois arrays de 20 números inteiros aleatórios (ex: de 1 a 50)
u = np.random.randint(1, 50, size=20)
v = np.random.randint(1, 50, size=20)

# Calcula o produto escalar
produto_escalar = np.dot(u, v)

# Calcula a norma (comprimento) de cada vetor
norma_u = np.linalg.norm(u)
norma_v = np.linalg.norm(v)

# Produto escalar normalizado (cosseno do ângulo)
# np.clip previne erros de precisão numéricas fora do intervalo [-1, 1]
cos_theta = np.clip(produto_escalar / (norma_u * norma_v), -1.0, 1.0)

# Calcula o ângulo em radianos e converte para graus
angulo_rad = np.arccos(cos_theta)
angulo_graus = np.degrees(angulo_rad)

print("Vetor u:\n", u)
print("\nVetor v:\n", v)
print(f"\nCosseno do ângulo: {cos_theta:.4f}")
print(f"Ângulo em Radianos: {angulo_rad:.4f} rad")
print(f"Ângulo em Graus: {angulo_graus:.2f}°")
```

```python
# 1

import numpy as np
from scipy import integrate

# 1. Definimos a função matemática: f(x) = x^2 * sin(x)
def f(x):
    return (x**2) * np.sin(x)

# 2. Definimos os limites de integração (de 0 até π)
limite_inferior = 0
limite_superior = np.pi

# 3. Calculamos a integral usando integrate.quad()
resultado, erro_estimado = integrate.quad(f, limite_inferior, limite_superior)

print(f"O valor da integral é aproximadamente: {resultado:.4f}")
print(f"Erro estimado do cálculo: {erro_estimado:.2e}")
```

```python
 # 2

import numpy as np
from scipy.integrate import solve_ivp

# 1. Definimos a função da EDO: dy/dx = f(x, y)
def derivada(x, y):
    return x - y

# 2. Condição inicial: y(0) = 1 (passada como uma lista ou array)
y0 = [1]

# 3. Intervalo de integração: de x = 0 até x = 10
intervalo_x = (0, 10)

# 4. Pontos em x onde queremos calcular o resultado
x_pontos = np.linspace(0, 10, 100)

# 5. Resolvemos a EDO usando o solve_ivp
solucao = solve_ivp(derivada, intervalo_x, y0, t_eval=x_pontos)

print("Status da solução:", solucao.message)
print("\nPrimeiros 5 valores de x:")
print(solucao.t[:5])

print("\nPrimeiros 5 valores correspondentes de y:")
print(solucao.y[0][:5])
```

```python
# 3

import numpy as np
from scipy.optimize import fsolve

# 1. Definimos a função f(x) = x^4 - 3x^3 + 2
def f(x):
    return x**4 - 3*x**3 + 2

# 2. Damos chutes iniciais para encontrar as raízes reais
# Chute para a primeira raiz real (perto de 1)
raiz1 = fsolve(f, x0=1)

# Chute para a segunda raiz real (perto de 3)
raiz2 = fsolve(f, x0=3)

print(f"Primeira raiz real: {raiz1[0]:.4f}")
print(f"Segunda raiz real:  {raiz2[0]:.4f}")
```

```python
# 4

import numpy as np
from scipy.optimize import minimize

# 1. Definimos a função NEGATIVA para o SciPy encontrar o mínimo dela
# f(x) = -x^2 + 4x + 1  =>  multiplicamos tudo por (-1)
def funcao_negativa(x):
    return -(-x**2 + 4*x + 1)

# 2. Damos um chute inicial para o ponto x (ex: x = 0)
chute_inicial = [0]

# 3. Usamos a função minimize do SciPy
resultado = minimize(funcao_negativa, chute_inicial)

# 4. Extraímos o ponto onde x é máximo e o valor de f(x)
x_max = resultado.x[0]
y_max = -resultado.fun  # Invertemos o sinal de volta para ter o valor máximo correto

print(f"O valor de x que maximiza a função é: x = {x_max:.2f}")
print(f"O valor máximo da função é: f(x) = {y_max:.2f}")
```

```python
# 5

import numpy as np
from scipy.fft import fft, fftfreq

# 1. Definimos o tempo de amostragem
taxa_amostragem = 100  # 100 amostras por segundo
duracao = 5            # 5 segundos de sinal
N = taxa_amostragem * duracao  # Total de pontos (500)

# Vetor de tempo t
t = np.linspace(0, duracao, N, endpoint=False)

# 2. Criamos o sinal composto: sin(2*pi*t) + cos(4*pi*t)
sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)

# 3. Calculamos a Transformada de Fourier (FFT)
fft_resultado = fft(sinal)

# 4. Calculamos o eixo de frequências correspondente
frequencias = fftfreq(N, 1 / taxa_amostragem)

# Como a FFT gera frequências espelhadas (positivas e negativas),
# pegamos apenas a metade positiva para visualizar:
metade = N // 2
freq_positivas = frequencias[:metade]
amplitudes = np.abs(fft_resultado[:metade]) * (2 / N) # Normalizamos a amplitude

# Imprime os picos detectados
picos = freq_positivas[amplitudes > 0.5]
print("Frequências detectadas no sinal (em Hz):", np.round(picos, 2))
```

```python
# 9

import numpy as np
from scipy.optimize import curve_fit

# 1. Definimos os pontos x e y dados no exercício
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])

# 2. Criamos o modelo matemático da função quadrática que queremos ajustar
def funcao_quadratica(x, a, b, c):
    return a * (x**2) + b * x + c

# 3. Executamos o curve_fit
# Ele devolve dois valores: os parâmetros (popt) e a matriz de covariância (pcov)
parametros, _ = curve_fit(funcao_quadratica, x, y)

# Extraímos os coeficientes a, b e c
a, b, c = parametros

print(f"Coeficientes encontrados:")
print(f"a = {a:.4f}")
print(f"b = {b:.4f}")
print(f"c = {c:.4f}")
print(f"\nA equação ajustada é: f(x) = {a:.4f}x² + {b:.4f}x + {c:.4f}")
```

```python
# 10

import numpy as np
from scipy import integrate

# 1. Definimos a função f(y, x)
# ATENÇÃO: No scipy.integrate.dblquad, a ordem dos parâmetros DEVE SER (y, x)
def f(y, x):
    return x * y + x**2

# 2. Limites de integração
# x vai de 0 a 2
limite_inf_x = 0
limite_sup_x = 2

# y vai de 0 a 1 (como são constantes, passamos como funções que retornam 0 e 1)
limite_inf_y = lambda x: 0
limite_sup_y = lambda x: 1

# 3. Calculamos a integral dupla
resultado, erro = integrate.dblquad(
    f,
    limite_inf_x,
    limite_sup_x,
    limite_inf_y,
    limite_sup_y
)

print(f"O valor da integral dupla é: {resultado:.4f}")
print(f"Erro estimado: {erro:.2e}")
```

```python
# 14

import numpy as np
from scipy.optimize import minimize

# 1. Definimos a função multivariada
# A variável v é um vetor onde: v[0] é o x e v[1] é o y
def f(v):
    x = v[0]
    y = v[1]
    return (x - 2)**2 + (y - 3)**2

# 2. Definimos um chute inicial para (x, y)
chute_inicial = [0, 0]

# 3. Minimizamos a função usando o scipy.optimize.minimize
resultado = minimize(f, chute_inicial)

# Extraímos os valores encontrados de x e y
x_min, y_min = resultado.x
valor_minimo = resultado.fun

print(f"Ponto de mínimo encontrado:")
print(f"x = {x_min:.4f}")
print(f"y = {y_min:.4f}")
print(f"Valor mínimo da função f(x, y) = {valor_minimo:.4f}")
```

```python
# 16

import numpy as np
from scipy.integrate import solve_ivp

# 1. Definimos o sistema de EDOs de 1ª ordem
# Y é uma lista/array com [y, y']
def sistema_edos(x, Y):
    y1, y2 = Y
    dy1_dx = y2
    dy2_dx = -y1
    return [dy1_dx, dy2_dx]

# 2. Condições iniciais: y(0) = 1 e y'(0) = 0
condicoes_iniciais = [1, 0]

# 3. Intervalo de integração (de x = 0 até x = 2π)
intervalo_x = (0, 2 * np.pi)

# 4. Pontos em x para avaliar a solução
x_pontos = np.linspace(0, 2 * np.pi, 100)

# 5. Resolvemos o sistema com solve_ivp
solucao = solve_ivp(sistema_edos, intervalo_x, condicoes_iniciais, t_eval=x_pontos)

# Extraímos os resultados de y(x) e y'(x)
x_valores = solucao.t
y_valores = solucao.y[0]    # Pega y(x)
y_linha_valores = solucao.y[1] # Pega y'(x)

print("Primeiros 5 valores calculados para y(x):")
print(np.round(y_valores[:5], 4))
```

```python
# 18

import numpy as np
from scipy.interpolate import CubicSpline

# 1. Definimos o conjunto de pontos x e calculamos y = x^2
x = np.array([1, 2, 3, 4, 5])
y = x**2  # y = [1, 4, 9, 16, 25]

# 2. Criamos a spline cúbica com os pontos dados
cs = CubicSpline(x, y)

# 3. Testamos a interpolação em novos valores intermediários de x (ex: 1.5, 2.5, 3.5, 4.5)
x_testes = np.array([1.5, 2.5, 3.5, 4.5])
y_interpolados = cs(x_testes)

# Calculamos os valores reais de x^2 para comparar com o resultado interpolado
y_reais = x_testes**2

print("Comparação entre os valores interpolados e os valores reais de x²:")
for xi, yi, yr in zip(x_testes, y_interpolados, y_reais):
    print(f"  x = {xi:.1f} | Interpolado: {yi:.4f} | Real (x²): {yr:.4f}")
```

```python
# 19

import numpy as np
from scipy.fft import fft, fftfreq
from scipy.signal import find_peaks

# 1. Sinal discreto (5 Hz e 12 Hz)
fs, t = 100, np.linspace(0, 2, 200, endpoint=False)
sinal = np.sin(2*np.pi*5*t) + np.cos(2*np.pi*12*t)

# 2. Fourier e frequências em 2 linhas
freqs = fftfreq(len(sinal), 1/fs)
amplitudes = np.abs(fft(sinal))

# 3. Acha os índices dos picos automaticamente
indices_picos, _ = find_peaks(amplitudes[:len(sinal)//2], height=10)

# Imprime os picos encontrados
print("Frequências encontradas (Hz):", freqs[indices_picos])
```

```python
# 20

import numpy as np
from scipy import linalg

# 1. Matriz de coeficientes (A)
A = np.array([
    [2, 1, 1],
    [1, 3, 2],
    [1, 0, 1]
])

# 2. Vetor dos resultados/constantes (B)
B = np.array([8, 13, 4])

# 3. Resolve o sistema A * X = B
X = linalg.solve(A, B)

# Extraímos os valores de x, y, z
x, y, z = X

print(f"Solução do Sistema:")
print(f"x = {x:.2f}")
print(f"y = {y:.2f}")
print(f"z = {z:.2f}")
```

```python
# 8

import matplotlib.pyplot as plt
import numpy as np

# 1. Preparação dos dados da função seno
x = np.linspace(0, 10, 100)
y = np.sin(x)

# 2. Definição do intervalo de confiança (desvio/margem)
# Criamos limites superior e inferior (ex: variação de ±0.2 ao redor de sin(x))
limite_inferior = y - 0.2
limite_superior = y + 0.2

# 3. Criação do gráfico
plt.figure(figsize=(9, 5))

# Desenha a linha principal
plt.plot(x, y, color='blue', linewidth=2, label='y = sin(x)')

# Aplica o sombreamento entre o limite inferior e superior
plt.fill_between(x, limite_inferior, limite_superior, color='blue', alpha=0.2, label='Intervalo de Confiança')

# 4. Personalizações
plt.title('Gráfico de $y = \sin(x)$ com Sombreamento de Confiança', fontsize=13)
plt.xlabel('Eixo X', fontsize=11)
plt.ylabel('Eixo Y', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='upper right')

# 5. Exibe o gráfico
plt.show()
```

```python
# 10

import matplotlib.pyplot as plt

# 1. Dados reais (sem matemática/fórmulas)
meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun']
vendas_loja_a = [10, 15, 12, 18, 22, 25]
vendas_loja_b = [8, 9, 14, 15, 13, 19]
vendas_loja_c = [5, 11, 8, 10, 16, 20]

# 2. Três linhas com cores e estilos diferentes
plt.plot(meses, vendas_loja_a, color='blue', linestyle='-', label='Loja A')    # Linha contínua
plt.plot(meses, vendas_loja_b, color='red', linestyle='--', label='Loja B')   # Linha tracejada
plt.plot(meses, vendas_loja_c, color='green', linestyle=':', label='Loja C')  # Linha pontilhada

# 3. Identificação do gráfico
plt.title('Vendas Mensais por Loja (2024)')
plt.xlabel('Mês')
plt.ylabel('Vendas (em mil R$)')
plt.grid(True)
plt.legend()

# 4. Exibir
plt.show()
```

```python
# 11

import matplotlib.pyplot as plt
import numpy as np

# 1. Categorias (Produtos) e Prazos de Validade (em meses)
produtos = ['Base', 'Batom', 'Rímel']

# Três grupos de tempo de validade (em meses):
fechado = np.array([36, 36, 24])       # Tempo guardado antes de abrir
apos_aberto = np.array([12, 18, 6])    # Tempo útil de uso após aberto
com_conservante = np.array([6, 6, 3])  # Extensão estimativa com boa conservação

# 2. Plot das barras empilhadas
plt.bar(produtos, fechado, label='Fechado (Validade original)', color='plum')
plt.bar(produtos, apos_aberto, bottom=fechado, label='Após Aberto', color='mediumpurple')
plt.bar(produtos, com_conservante, bottom=fechado + apos_aberto, label='Margem de Segurança', color='indigo')

# 3. Identificação do gráfico
plt.title('Validade Total dos Produtos de Maquiagem')
plt.xlabel('Produto')
plt.ylabel('Tempo de Validade (Meses)')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.legend()

# 4. Exibir
plt.show()
```

```python
# 12

import matplotlib.pyplot as plt
import numpy as np

# 1. Dados: Função quadrática y = x²
x = np.linspace(-10, 10, 200)
y = x**2

# 2. Plot da linha
plt.plot(x, y, color='purple', linewidth=2)

# 3. Ajuste dos limites dos eixos X e Y
plt.xlim(-5, 5)   # Foca no intervalo de x = -5 a x = 5
plt.ylim(0, 25)   # Ajusta o eixo Y correspondente aos valores de x²

# 4. Título, rótulos e grade
plt.title('Gráfico da Função $y = x^2$')
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')
plt.grid(True)

# 5. Exibir
plt.show()
```

```python
# 13

import matplotlib.pyplot as plt
import numpy as np

# 1. Gera 100 pontos aleatórios para as coordenadas X, Y e Z
np.random.seed(42)
x = np.random.rand(100) * 100
y = np.random.rand(100) * 100
z = np.random.rand(100) * 100

# 2. Cria a figura e ativa o modo 3D
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

# 3. Plota os pontos em 3D
ax.scatter(x, y, z, c=z, cmap='viridis', s=50, alpha=0.8)

# 4. Título e rótulos dos três eixos
ax.set_title('Gráfico de Dispersão 3D (100 Pontos)')
ax.set_xlabel('Eixo X')
ax.set_ylabel('Eixo Y')
ax.set_zlabel('Eixo Z')

# 5. Exibe o gráfico
plt.show()
```

```python
# 16

import matplotlib.pyplot as plt
import numpy as np

# 1. Dados: Aquecimento da água ao longo do tempo (em minutos)
tempo = np.linspace(0, 10, 100)
# A temperatura vai de -20°C (gelo) até 100°C (ebulição)
temperatura = -20 + 12 * tempo

# 2. Plot do gráfico de linha
plt.figure(figsize=(8, 4))
plt.plot(tempo, temperatura, color='firebrick', linewidth=2)

# 3. Títulos para os eixos e o gráfico
plt.title('Aquecimento da Água ao Longo do Tempo')
plt.xlabel('Tempo (minutos)')
plt.ylabel('Temperatura (°C)')

# 4. Ponto x = 5 (momento em que a água atinge 40°C)
temp_em_5min = -20 + 12 * 5  # Temperatura no minuto 5 (40°C)

# Marcação do ponto com uma bolinha vermelha
plt.plot(5, temp_em_5min, 'ro')

# Rótulo de texto no ponto x = 5
plt.text(5.2, temp_em_5min - 2, 'Ebulição (40°C)', fontsize=10, fontweight='bold', color='darkred')

# 5. Grade e exibição
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

plt.savefig('grafico_aquecimento.pdf', bbox_inches='tight')
print('Gráfico salvo com sucesso!')
```

```python
# 18

import matplotlib.pyplot as plt
import numpy as np

# 1. Dados: Categorias e valores (positivos e negativos)
categorias = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun']
valores = np.array([150, -80, 200, -120, 50, -30])

# 2. Definição de cores
cores = ['mediumseagreen' if v > 0 else 'crimson' for v in valores]

# 3. Plot do gráfico de barras
plt.figure(figsize=(8, 4))
plt.bar(categorias, valores, color=cores)

# 4. Linha de referência no zero (eixo neutro)
plt.axhline(0, color='black', linewidth=1)

# 5. Títulos e rótulos
plt.title('Balanço Financeiro Semestral (Lucro vs Prejuízo)')
plt.xlabel('Mês')
plt.ylabel('Valor (R$)')
plt.grid(axis='y', linestyle='--', alpha=0.5)

# 6. Exibir
plt.show()
```

```python
# 19

import matplotlib.pyplot as plt
import numpy as np

# 1. Dados: Duas séries ao longo do tempo (exemplo de temperatura máxima e mínima)
dias = ['Seg', 'Ter', 'Qua', 'Qui', 'Sex', 'Sáb', 'Dom']
temp_max = [28, 30, 25, 22, 27, 31, 29]
temp_min = [18, 19, 15, 14, 16, 20, 18]

# 2. Plot das duas séries de linhas
plt.figure(figsize=(8, 4))
plt.plot(dias, temp_max, color='firebrick', marker='o', linewidth=2, label='Temp. Máxima')
plt.plot(dias, temp_min, color='royalblue', marker='s', linewidth=2, label='Temp. Mínima')

# 3. Título e rótulos dos eixos
plt.title('Previsão do Tempo Semanal')
plt.xlabel('Dia da Semana')
plt.ylabel('Temperatura (°C)')
plt.grid(True, linestyle='--', alpha=0.5)

# 4. Adiciona a LEGENDA para diferenciar as duas séries
plt.legend(loc='upper right')

# 5. Exibir
plt.show()
```

```python
# 20

import matplotlib.pyplot as plt
import numpy as np

# 1. Dados: y = x^2 com x variando de 0 a 10
x = np.linspace(0, 10, 100)
y = x ** 2

# 2. Plot do gráfico de linha
plt.figure(figsize=(8, 4))
plt.plot(x, y, color='darkviolet', linewidth=2, label='$y = x^2$')

# 3. Título e rótulos
plt.title('Gráfico de $y = x^2$ com Grade Personalizada')
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')

# 4. Define o espaçamento dos eixos (Ticks)
# Eixo X vai de 0 a 10 com marcas a cada 1 unidade
plt.xticks(np.arange(0, 11, 1))

# Eixo Y vai de 0 a 100 com marcas a cada 10 unidades
plt.yticks(np.arange(0, 101, 10))

# 5. Adiciona a grade personalizada acompanhando as marcas definidas
plt.grid(True, which='both', linestyle='--', linewidth=0.7, color='gray', alpha=0.6)

plt.legend()
plt.show()
```
