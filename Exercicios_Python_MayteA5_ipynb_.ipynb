```python
# 20
"""Neste exercício, criamos uma matriz 2x3 usando uma lista de listas.
Em seguida, calculamos a matriz transposta. A transposta é obtida invertendo
as linhas e colunas da matriz original. Para isso, utilizamos uma compreensão de lista
aninhada, onde os índices 'i' e 'j' são trocados para construir a nova matriz.

- for i in range(len(matriz))], o for ler todos os elemnetos da linha.
- for j in range(len(matriz[0]))], o for ler todos os elemnetos da coluna.
- matriz[i][j] acessa o elemento na posição '(i, j)' e o coloca na nova posição '(j, i)'""".


matriz = [[1,2,3], [3,2,1]]

transposta = [[matriz[i][j] for i in range(len(matriz))] for j in range(len(matriz[0]))]

print("A matriz é:", matriz)
print("A transposta é:", transposta)
```

```python
# 21
"""Como a busca binária divide pela metade uma lista e comparando esse meio com alvo, eu criei um def que guardar
uma função que recebe como parâmetro uma lista e um alvo. A esquerda seria a esquerda da lista e a direita da lista, uma
comecei com 0 e a outra usei o tamanho da lista com 'len'. Dentro do while comparei esquerda e direita e depois dividir para
achar o meio. Depois fiz condições para achar o alvo. E no final usei o def na lista que declarei."""

def busca_binaria(lista, alvo):
  esquerda = 0
  direita = len(lista) - 1
  while esquerda <= direita:
    meio = (esquerda + direita) // 2
    if lista[meio] == alvo:
      return meio
    elif lista[meio] < alvo:
      esquerda = meio + 1
    else:
      direita = meio - 1
  return "Alvo não encontrado"

lista = [1,2,3,4,5,6,7]
alvo = 3
resultado = busca_binaria(lista, alvo)
print(f"O alvo foi encontrado {resultado}")
```

```python
# 22
"""Aqui a variável converter recebe a palavra criptografada, o for acha cada caracter da mensagem
e se estiver no alfabeto o 'find' encontra a posicao do caracter, desse modo achamos o caracter e somamamos
mais tres casas, usando o limite da lista com 'len(caracter)', logo depois acrescentamos a nova letra e guardamos
em converter"""

mensagem = "O rato correu do gato"
posicao = 3

alfabeto = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
converter = ""
mensagem = mensagem.upper()

for caracter in mensagem:
  if caracter in alfabeto:
    num = alfabeto.find(caracter)
    num = (num + posicao) % len(alfabeto)
    converter += alfabeto[num]
  else:
    converter = converter + caracter

print(f"A mensagem criptografada é: {converter}")
```

```python
# 23
"""Aqui a variável converter recebe a palavra  descriptografda, o for acha cada caracter da mensagem
e se estiver no alfabeto o 'find' encontra a posicao do caracter, desse modo achamos o caracter e subtraimos
tres casas, usando o limite da lista com 'len(caracter)', logo depois acrescentamos a nova letra e guardamos
em converter"""

mensagem = "R UDWR FRUUHX GR JDWR"
posicao = 3

alfabeto = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
converter = ""
mensagem = mensagem.upper()

for caracter in mensagem:
  if caracter in alfabeto:
    num = alfabeto.find(caracter)
    num = (num - posicao) % len(alfabeto)
    converter += alfabeto[num]
  else:
    converter = converter + caracter

print(f"A mensagem descriptografada é: {converter}")
```

```python
# 24
"""Este excecício verifica a validade de um CPF com base em suas regras de formação,
especificamente os dois dígitos verificadores. Na paret que o numero entra o 'isdigt' identifica os digitos, logo depois
verifica se tem 11 digitos, ai entramos na validação, na qual soma_1 os digitos seram multiplicados de 2 ao 10 e com sum vão
ser somados. Com isso descobrimos o resto_1 e fazemos a comparação de valadação de cpf, se for menor que 2 é 0 se for maior é
resto menos 11. No final comparamso os digitos verificados com os dois digitos verificadores"""

cpf_input = input("Digite o CPF (apenas números): ")

numeros = [int(digito) for digito in cpf_input if digito.isdigit()]

if len(numeros) != 11:
    print("CPF Inválido: Deve conter 11 dígitos.")
else:
    # Cálculo do 1º Dígito Verificador
    soma_1 = sum(numeros[i] * peso for i, peso in enumerate(range(10, 1, -1)))
    resto_1 = soma_1 % 11
    digito_1 = 0 if resto_1 < 2 else 11 - resto_1

    # Cálculo do 2º Dígito Verificador
    # Para o segundo dígito, inclui-se o primeiro dígito verificador na soma com peso 11.
    # Range(11, 1, -1) é (11, 10, 9, ..., 2)
    soma_2 = sum(numeros[i] * peso for i, peso in enumerate(range(11, 1, -1)))
    resto_2 = soma_2 % 11
    digito_2 = 0 if resto_2 < 2 else 11 - resto_2

    # Comparação dos Dígitos
    if digito_1 == numeros[9] and digito_2 == numeros[10]:
        print("CPF válido!")
    else:
        print("CPF inválido!")
```

```python

# 30

import random

# Nessa parte usei o 'random.choice', pois ele registra um elemento aleátorio da lista
# Na variável letras_descorbetas  o '_' guarda as letras encontradas em palavras_secretas
def jogar():
    palavras = ["PYTHON", "PROGRAMACAO", "COMPUTADOR", "DESENVOLVEDOR"]
    palavra_secreta = random.choice(palavras)
    letras_descobertas = ["_" for _ in palavra_secreta]

    # Temos uma lista que guarda as letras_tentadas
    chances = 6
    erros = 0
    letras_tentadas = []

    print("Bem-vindo ao Jogo da Forca!")

    # Ele faz o jogando continuar rodando enquanto o número de erro for menor que os de chances
    # E o '.join' atualiza as letras esncontradas adicionando espaço
    while erros < chances and "_" in letras_descobertas:
        print("\nPalavra: ", " ".join(letras_descobertas))
        print(f"Erros: {erros}/{chances}")

        chute = input("Digite uma letra: ").strip().upper()

        # Aqui impede que eltras sejam repetidas comparando as que ja foram adicionadas em letras_tentadas
        if chute in letras_tentadas:
            print("Você já tentou esta letra.")
            continue

        letras_tentadas.append(chute)

        # Nessa parte o enumerate verifica se dentro da palavra secreta a letra é verdadeira
        # e a condicional adiciona ela em letras descobertas
        if chute in palavra_secreta:
            for i, letra in enumerate(palavra_secreta):
                if letra == chute:
                  letras_descobertas[i] = chute
        # Contagem de erros
        else:
            erros += 1
            print(f"Letra incorreta!")

        if "_" not in letras_descobertas:
            print(f"\nParabéns, você ganhou! A palavra era: {palavra_secreta}")
        elif erros == chances:
            print(f"\nFim de jogo! Você perdeu. A palavra era: {palavra_secreta}")
# Inicializa o jogo
if __name__ == "__main__":
    jogar()
```

```python
# 31
"""Nesse excício, o for pecorre cada elemento da lista comparando se o
elemento j é maior que seu próximo. Se for trocam de lugar, até pecorrer
toda a lista."""
lista = [23, 2, 5, 80, 76]
n = len(lista)

for i in range(n):
    for j in range(0, n - i - 1):
       if lista[j] > lista[j + 1]:
           lista[j], lista[j + 1] = lista[j + 1], lista[j]

print("Lista ordenada:", lista)
```

```python
# 35
"""Neste exercício, o programa calcula a sequência de Fibonacci até um número 'n' fornecido pelo usuário.
As variáveis 'a' e 'b' são inicializadas como os dois primeiros números da sequência (0 e 1).
O loop 'while' continua adicionando 'a' na lista 'fibonacci' e atualizando 'a' e 'b' (onde 'a' recebe o valor anterior de 'b', e 'b' recebe a soma de 'a' e 'b')
até que 'a' seja maior que 'n'."""

n = int(input("Digite o valor até onde deseja que o fibonacci pare:"))

a, b = 0, 1
fibonacci = []

while a <= n:
    fibonacci.append(a)
    a, b = b, a + b

print(f"Sequência até o valor {n}:", fibonacci)
```

```python
# 36
"""Neste exercício, o programa gera uma senha aleatória de um determinado comprimento, usamos de diferente o 'random.choice' que escolhi
caracter aleátoriamente que é somado na variável senha. Concomitantemente pedimos o tamanho da senha, gerando uma senha sorteada."""
import random

def gera_senha(tamanho):
    caracteres = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789"
    senha = ""
    for i in range(tamanho):
        senha += random.choice(caracteres)
    return senha
# Pedi pro usuário o tamanho que ele deseja para a senha
tamanho_senha = int(input("Digite o tamanho da senha desejada: "))
# Chamei a função e exibo a senha gerada na tela
senha_gerada = gera_senha(tamanho_senha)
print(f"Sua senha é: {senha_gerada}")
```

```python
# 37
"""Neste exercício, o programa calcula o fatorial de um número inteiro usando uma função recursiva.
A função 'fatorial(n)' verifica se 'n' é 0 ou 1, retornando 1.
Caso contrário, ela chama a si mesma (n * fatorial(n - 1)) para multiplicar o número atual pelo fatorial do número anterior, até atingir o caso base.
O programa solicita ao usuário um número, chama a função fatorial e exibe o resultado."""

def fatorial(n):
  if n == 0 or n == 1:
    return 1
  else:
    return n * fatorial(n - 1)

numero = int(input("Digite um número para calcular o fatorial: "))
resultado = fatorial(numero)
print(f"O fatorial de {numero} é {resultado}")
```

```python
# 40

lista_compras = []

print("Bem vindo ao gerenciador de lista de compras!")

def exibir_menu():
  print("\n--- MENU ---")
  print("1 - ADICIONAR item(ns)")
  print("2 - REMOVER item")
  print("3 - LISTAR itens")
  print("4 - SAIR")
  # Valida a entrada e verifica se está entre o valor de 1-4, caso de erro o programa não para por causa try e except
  while True:
    escolha = input("Digite sua opção (1-4): ").strip()
    try:
      num_escolha = int(escolha)
      return num_escolha
    except ValueError:
      print("Entrada inválida. Por favor, digite um número de 1 a 4.")
# Enquanto o usúario digitar para sair o programa continua
while True:
  opcoes = exibir_menu()
  """Aqui usei strip para separar a palavras e split para separar por vírgula, o for verifica se
  existe item e adiciona a lista"""
  match opcoes:
    case 1:
      entrada_itens = input("O que deseja adicionar? (Separe por vírgulas se for mais de um item) ")
      itens_a_adicionar = [item.strip() for item in entrada_itens.split(',') if item.strip()]
      for item in itens_a_adicionar:
        lista_compras.append(item)
        print(f"'{item}' adicionado à lista.")
    # Aqui verifica e remove
    case 2:
      item_a_remover = input("O que deseja remover? ").strip()
      if item_a_remover in lista_compras:
        lista_compras.remove(item_a_remover)
        print(f"'{item_a_remover}' removido da lista.")
      else:
        print(f"'{item_a_remover}' não encontrado na lista.")
    case 3:
      # Aqui o enumerate pega a posição e o valor do alimento na lista e soma na lista
      print("\n--- Sua Lista de Compras ---")
      if not lista_compras:
        print("Sua lista está vazia.")
      else:
        for i, item in enumerate(lista_compras):
          print(f"{i+1}. {item}")
    case 4:
      print("Obrigado por usar a lista de compras!")
      break
    case _:
      print("Opção inválida! Por favor, escolha um número entre 1 e 4.")
```

```python
# 43
"""Neste exercício, criei um conversor de unidades de comprimento. O programa apresenta um menu de opções.
  O usuário escolhe uma opção e insere o valor a ser convertido.
  Utilizei a estrutura 'match/case' para lidar com as diferentes opções de conversão de forma mais limpa e eficiente,
  calculando e exibindo o resultado."""

print("CONVERSOR DE UNIDADES DE COMPRIMENTO")
print("1. Metros (m) para Centímetros (cm)")
print("2. Centímetros (cm) para Metros (m)")
print("3. Quilômetros (km) para Metros (m)")
print("4. Metros (m) para Quilômetros (km)")
print("5. Centímetros (cm) para Milímetros (mm)")
print("6. Milímetros (mm) para Centímetros (cm)")

opcao = input("\nEscolha a conversão desejada (1 a 6): ")

if opcao in ["1", "2", "3", "4", "5", "6"]:
    valor = float(input("Digite o valor a ser convertido: "))

    match opcao:
        case "1":
            resultado = valor * 100
            print(f"{valor} m = {resultado} cm")
        case "2":
            resultado = valor / 100
            print(f"{valor} cm = {resultado} m")
        case "3":
            resultado = valor * 1000
            print(f"{valor} km = {resultado} m")
        case "4":
            resultado = valor / 1000
            print(f"{valor} m = {resultado} km")
        case "5":
            resultado = valor * 10
            print(f"{valor} cm = {resultado} mm")
        case "6":
            resultado = valor / 10
            print(f"{valor} mm = {resultado} cm")
else:
    print("Opção inválida! Escolha um número de 1 a 6.")
```

```python
# 44
"""Neste exercício, a função 'primeiros_n_fibonacci(n)' gera os 'n' primeiros números da sequência de Fibonacci.
Ela inicializa a sequência com 0 e 1, e então usa um loop 'for' para adicionar os próximos números à lista, onde cada número é a soma dos dois anteriores.
O usuário informa quantos números da sequência deseja, e a função retorna e exibe essa lista."""

def primeiros_n_fibonacci(n):
    sequencia = []
    a, b = 0, 1
    for _ in range(n):
        sequencia.append(a)
        a, b = b, a + b
    return sequencia

n = int(input("Digite a posiçao desejada que o fibonnaci pare: "))
fibonacci = primeiros_n_fibonacci(n)
print(f"Os primeiros {n} números da sequência de Fibonacci são: {fibonacci}")
```

```python
# 45
"""Neste exercício, a função 'classificacao(numero)' recebe um número como entrada e retorna se ele é 'par' ou 'ímpar'.
Isso é feito usando o operador de módulo % para verificar se o resto da divisão do número por 2 é zero. Se for zero, o número é par; caso contrário, é ímpar.
Depois, uma lista de 'numeros' é iterada, e para cada número na lista, a função 'classificacao' é chamada para imprimir sua classificação."""
def classificacao(numero):
  if numero % 2 == 0:
    return "par"
  else:
    return "ímpar"

numeros = [10, 30, 5, 0, 8, 6]
for num in numeros:
  print(f"O número {num} é {classificacao(num)}")
```

```python
# 46
"""Neste exercício, o código calcula a média e o desvio padrão de uma lista de números.
Primeiramente, verifica se a lista está vazia para evitar erros de divisão.
Em seguida, calcula a soma total dos números e a quantidade de elementos para determinar a média.
Para o desvio padrão, calcula a soma dos quadrados das diferenças entre cada número e a média.
Finalmente, divide essa soma pela quantidade para obter a variância e, em seguida,
calcula a raiz quadrada da variância para encontrar o desvio padrão, exibindo todos os resultados formatados."""

import math

dados = [10, 12, 23, 23, 16, 23, 21, 16]

# Verifica se a lista não está vazia
if len(dados) == 0:
    print("A lista está vazia!")
else:
    # Cálculo da Média
    quantidade = len(dados)
    soma_total = sum(dados)
    media = soma_total / quantidade

    # Cálculo das diferenças ao quadrado em relação à média
    soma_diferencas_quadrado = 0
    for x in dados:
        diferenca = x - media
        soma_diferencas_quadrado += diferenca ** 2

    # Cálculo da Variância e do Desvio Padrão
    variancia = soma_diferencas_quadrado / quantidade
    desvio_padrao = math.sqrt(variancia)


    print(f"Lista de números: {dados}")
    print(f"Média: {media:.2f}")
    print(f"Desvio Padrão: {desvio_padrao:.2f}")
```

```python
# 47
"""Neste exercício, o código conta a frequência de cada número em uma lista. Ele inicializa um dicionário vazio 'contagem' para armazenar os números como chaves e suas respectivas ocorrências como valores.
Em seguida, itera sobre cada número na lista de entrada: se o número já estiver no dicionário, seu contador é incrementado; caso contrário, o número é adicionado ao dicionário com um contador inicial de 1.
Finalmente, a lista original e a frequência de cada número são exibidas de forma organizada."""

numeros = [4, 2, 8, 2, 4, 9, 2, 8, 4, 1]

contagem = {}
for num in numeros:
    if num in contagem:
        contagem[num] += 1
    else:
        contagem[num] = 1

print("Lista original:", numeros)
print("\nFrequência dos números:")

for num, qtd in contagem.items():
    print(f"O número {num} aparece {qtd} vezes.")
```

```python
# 48
"""Nesse exercício, o código implementa o algoritmo de ordenação por inserção (Insertion Sort) para organizar
uma lista de números. Ele percorre a lista, pegando cada elemento e inserindo-o na posição
correta na parte já ordenada da lista."""
def insertion_sort(lista):

    for i in range(1, len(lista)):
        chave = lista[i]
        j = i - 1
        while j >= 0 and lista[j] > chave:
            lista[j + 1] = lista[j]
            j -= 1

        lista[j + 1] = chave

    return lista

numeros = [12, 11, 13, 5, 6]
print("Lista desordenada:", numeros)
insertion_sort(numeros)
print("Lista ordenada:", numeros)
```
